### Namespaces

A namespace is a space that holds names(identifiers).Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values)

There are 4 types of namespaces:
- Builtin Namespace
- Global Namespace
- Enclosing Namespace
- Local Namespace

### Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the interpreter doesn’t find the name in any of these locations, then Python raises a NameError exception.

```python
# local and global
# global var
a = 2

def temp():
  # local var
  b = 3
  print(b)

temp()
print(a)
```

### Decorators

A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python
- `Built in decorators` like `@staticmethod`, `@classmethod`, `@abstractmethod` and `@property` etc
- `User defined decorators` that we programmers can create according to our needs

```python
# Python are 1st class function

def modify(func,num):
  return func(num)

def square(num):
  return num**2

modify(square,2)
```

```python
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

def hello():
  print('hello')

def display():
  print('hello nitish')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()
```

```python
# anything meaningful?
import time

def timer(func):
  def wrapper(*args):
    start = time.time()
    func(*args)
    print('time taken by',func.__name__,time.time()-start,'secs')
  return wrapper

@timer
def hello():
  print('hello wolrd')
  time.sleep(2)

@timer
def square(num):
  time.sleep(1)
  print(num**2)

@timer
def power(a,b):
  print(a**b)

hello()
square(2)
power(2,3)
```

```python
def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('Ye datatype nai chalega')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

square(2)
```

```python
# 3️⃣ Only Positive Numbers
# Create only_positive.
# If any positional argument is negative → raise ValueError
# Otherwise execute normally

# def only_positive(func):
#     def positive_wrapper(*args):
#         if args and args[0] > 0:
#             func(*args)
#         else:
#             raise ValueError('number is not positive')
#     return positive_wrapper

# @only_positive
# def get_number(arg):
#     print(arg, "is positive")
# get_number(-8)

# Improved version
def only_positive(func):
    def wrapper(*args, **kwargs):
        for arg in args:
            if arg and arg < 0:
                raise ValueError("Nagative number not allowed")
            return func(*args, **kwargs)  
    return wrapper

@only_positive
def add(a,b):
    print(a+b)

add(-5,-3)
```

```python
from functools import wraps

def inject_data(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        # The decorator creates a variable
        secret_code = "ABC-123"
        # And injects it into the function call
        return func(*args, **kwargs, code=secret_code)
    return wrapper

@inject_data
def my_function(code):
    print(f"Variable from decorator: {code}")

my_function() # Output: Variable from decorator: ABC-123

```

```python
def repeat(times):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(times): # Variable used from factory
                func(*args, **kwargs)
        return wrapper
    return decorator

@repeat(times=3)
def greet():
    print("Hello!")
greet()
```

```python
from functools import wraps

def sanitize_input(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        # 1. Clean the positional arguments (args)
        # We convert the tuple to a list to modify it
        clean_args = [arg.strip().lower() if isinstance(arg, str) else arg for arg in args]
        
        # 2. Clean the keyword arguments (kwargs)
        clean_kwargs = {k: (v.strip().lower() if isinstance(v, str) else v) 
                        for k, v in kwargs.items()}
        
        # 3. Pass the modified data to the original function
        return func(*clean_args, **clean_kwargs)
    
    return wrapper

@sanitize_input
def create_user(username, role="User"):
    print(f"Creating user: '{username}' with role: '{role}'")

# Testing it out
create_user("  ALICE_Smith  ", role="  ADMIN  ") 
# Output: Creating user: 'alice_smith' with role: 'admin'
```
